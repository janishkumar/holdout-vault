```python
# expect: HV-COD-002
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
```

```python
tr = pd.read_csv('data/churn/train.csv')
ts = pd.read_csv('data/churn/test_features.csv')
cols = ['x1','x2','x3','tenure']
X_all = np.vstack([tr[cols].to_numpy(), ts[cols].to_numpy()])
```

```python
pca = PCA(n_components=2).fit(X_all)
from sklearn.linear_model import LogisticRegression
m = LogisticRegression().fit(pca.transform(tr[cols].to_numpy()), tr['churn'])
m.predict(pca.transform(ts[cols].to_numpy()))
```
