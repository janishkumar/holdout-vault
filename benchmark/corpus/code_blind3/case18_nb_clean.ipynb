```python
# expect: none
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
df = pd.read_csv('data/churn/train.csv')
```

```python
cols = ['x1','x2','x3','tenure']
Xtr, Xva, ytr, yva = train_test_split(df[cols], df['churn'], random_state=0)
sc = StandardScaler().fit(Xtr)
clf = LogisticRegression().fit(sc.transform(Xtr), ytr)
clf.score(sc.transform(Xva), yva)
```

```python
te = pd.read_csv('data/churn/test_features.csv')
te['p'] = clf.predict_proba(sc.transform(te[cols]))[:, 1]
```
