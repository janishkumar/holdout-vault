```python
# expect: HV-COD-001, HV-COD-005
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
df = pd.read_csv('data/churn/train.csv')
cols = ['x1','x2','x3','tenure']
```

```python
Xr, yr = SMOTE(random_state=0).fit_resample(df[cols], df['churn'])
Xs = StandardScaler().fit_transform(Xr)
```

```python
Xtr, Xva, ytr, yva = train_test_split(Xs, yr, random_state=0)
RandomForestClassifier().fit(Xtr, ytr).score(Xva, yva)
```
