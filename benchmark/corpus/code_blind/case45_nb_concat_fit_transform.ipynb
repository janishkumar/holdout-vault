```python
# expect: HV-COD-002
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
train = pd.read_csv('data/churn/train.csv')
test = pd.read_csv('data/churn/test_features.csv')
num = ['x1','x2','x3']
all_df = pd.concat([train[num], test[num]])
```

```python
sc = StandardScaler()
Z = sc.fit_transform(all_df)
clf = LogisticRegression().fit(Z[:len(train)], train['churn'])
clf.predict(Z[len(train):])
```
